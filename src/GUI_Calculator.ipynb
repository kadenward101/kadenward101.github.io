```python
from tkinter import *
```

```python
def button_click(value):
    current = entry.get()
    entry.delete(0, END)               #clearing the entry in the text field 
    entry.insert(0, current + value)
```

```python
def clear():
    entry.delete(0, END)
```

```python
def evaluate():                             #evaluates the string value entered in the field 
    try:
        expression = entry.get()            #evaluates the string value entered in the input field 
        result = eval(expression)           #used to retrieve the string which is entered by the user
        entry.delete(0, END)
        entry.insert(0, str(result))        #inserts the result from the input string
    except Exception:
        entry.delete(0, END)
        entry.insert(0, "Error")            #clears the field and fills the field with the string "Error" to show mathematical errors
```

```python
def new_tab():
    new_window = Toplevel(window)           #Toplevel(window) is used to create an addition window to the already open application 
    new_window.title("New Tab")
    new_window.geometry("300x200")
```

```python
def new_window():
    new_win = Toplevel(window)
    new_win.title("New Window")
    new_win.geometry("500x400")
```

```python
window = Tk()
window.geometry("325x400")
window.title("Calculator")
window.configure(bg="lime")  #used to adjust the color of the background number display screen 
```

```python
menu_bar = Menu(window)
file_menu = Menu(menu_bar, tearoff=0)
edit_menu = Menu(menu_bar, tearoff=0)
view_menu = Menu(menu_bar, tearoff=0)

menu_bar.add_cascade(label="File", menu=file_menu)
file_menu.add_command(label="New Tab", command=new_tab)
file_menu.add_command(label="New Window", command=new_window)
file_menu.add_command(label="Open")
file_menu.add_command(label="Exit", command=window.quit)

menu_bar.add_cascade(label="Edit", menu=edit_menu)
edit_menu.add_command(label="Cut")
edit_menu.add_command(label="Copy")
edit_menu.add_command(label="Paste")
edit_menu.add_command(label="Delete")

menu_bar.add_cascade(label="View", menu=view_menu)
view_menu.add_command(label="Zoom")
view_menu.add_command(label="Status Bar")
view_menu.add_command(label="Word wrap")

window.config(menu=menu_bar)
```

```python
entry = Entry(window, width = 16, font=("Arial", 24), borderwidth=2, relief="solid", justify = "right", bg="white")
entry.grid(row = 0, column=0, columnspan=4, pady=10, sticky="nsew")
window.grid_rowconfigure(0, weight=2)
              
```

```python
'''justify = right allows the insert text pointer to be positioned on the right hand side of the entry field'''
```

```python
buttons = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('+', 4, 2), ('=', 4, 3),
    ('C', 5, 0, 4)  # 'C' for clear button

]
```

```python
for i in range(6):
    window.grid_rowconfigure(i, weight=1)
for i in range(4):
    window.grid_columnconfigure(i, weight=1)
```

```python
for (text, row, col, *span) in buttons:
    button = Button(window, text=text, font=("Arial", 18), width=5, height=2,
                    command=lambda t=text: button_click(t) if t not in ['=', 'C'] else evaluate() if t == '=' else clear())
    button.grid(row=row, column=col, columnspan=span[0] if span else 1, sticky="nsew")
   
```

```python
'''command = lambda... is used to capture and assign the action given by the button which pressed on the calculator

button.grid(row=row, column=col, columnspan=span[0] if span else 1, sticky="nsew") is used for the dynamic size adjustment'''
```

```python
window.mainloop()
```
